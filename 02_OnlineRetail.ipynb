```python
# TODO
df_clean['Sales'] = df_clean['Quantity'] * df_clean['UnitPrice']
cols = ['Quantity', 'UnitPrice', 'Sales']

# Range
range_vals = df_clean[cols].max() - df_clean[cols].min()
print(range_vals)
```

## Group 3 — Location and Shape

```python
# TODO
# vị trí
q1 = df_clean[cols].quantile(0.25)
q2 = df_clean[cols].quantile(0.50) # Tương đương Median
q3 = df_clean[cols].quantile(0.75)
```

```python
# hình dáng
skewness = df_clean[cols].skew()
kurtosis = df_clean[cols].kurt()

location_shape = pd.DataFrame({
    'Q1 (25%)': q1,
    'Q2 (50% - Median)': q2,
    'Q3 (75%)': q3,
    'Skewness (Độ lệch)': skewness,
    'Kurtosis (Độ nhọn)': kurtosis
})

print(location_shape)
```

## Câu hỏi 1: Quốc gia nào đóng góp doanh thu cao nhất, chiếm bao nhiêu % tổng doanh thu?

```python
# TODO
df['Sales'] = df['Quantity'] * df['UnitPrice']

country_sales = df.groupby('Country')['Sales'].sum()
total_sales = country_sales.sum()

country_summary = pd.DataFrame({
    'Doanh_thu': country_sales,
    'Ty_le_%': (country_sales / total_sales) * 100
}).sort_values(by='Doanh_thu', ascending=False)

top_country = country_summary.iloc[0]

print("QUỐC GIA CÓ DOANH THU CAO NHẤT")
print(f"Quốc gia: {country_summary.index[0]}")
print(f"Doanh thu: {top_country['Doanh_thu']:,.2f}")
print(f"Tỷ lệ chiếm: {top_country['Ty_le_%']:.2f}%")

print("\nTOP 5 QUỐC GIA")
print(country_summary.head())
```

## Câu hỏi 2: Sản phẩm nào bán chạy nhất theo doanh thu?

```python
# TODO
df['Sales'] = df['Quantity'] * df['UnitPrice']

product_sales = df.groupby(['StockCode', 'Description'])['Sales'].sum().reset_index()

top_products = product_sales.sort_values(by='Sales', ascending=False)

print("TOP 5 SẢN PHẨM CÓ DOANH THU CAO NHẤT (BAO GỒM PHÍ DỊCH VỤ)")
print(top_products.head(5).to_string(index=False))

non_products = ['DOT', 'POST', 'M', 'AMAZONFEE', 'BANK CHARGES', 'D']
filtered_products = top_products[~top_products['StockCode'].isin(non_products)]

print("\nTOP 5 MẶT HÀNG THƯƠNG MẠI CÓ DOANH THU CAO NHẤT")
print(filtered_products.head(5).to_string(index=False))
```
