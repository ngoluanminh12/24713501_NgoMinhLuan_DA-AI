## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
# TODO
survival = df.groupby('pclass')['survived'].mean()

print("--- TỶ LỆ SỐNG SÓT THEO HẠNG VÉ ---")
print(survival)

max_class = survival.idxmax()
max_rate = survival.max()

min_class = survival.idxmin()
min_rate = survival.min()

chenh_lech = max_rate - min_rate

print(f"\n- Hạng vé có tỷ lệ sống sót cao nhất: Hạng {max_class} ({max_rate:.2%})")
print(f"- Hạng vé có tỷ lệ sống sót thấp nhất: Hạng {min_class} ({min_rate:.2%})")
print(f"- Mức chênh lệch: {diff:.2%} (hoặc gấp {max_rate/min_rate:.2f} lần)")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO
sex_survival = df.groupby('sex')['survived'].mean()
sex_diff = sex_survival['female'] - sex_survival['male']

class_survival = df.groupby('pclass')['survived'].mean()
class_diff = class_survival[1] - class_survival[3]

print(f"Tỷ lệ sống theo Giới tính:\n{sex_survival}\nChênh lệch Nam - Nữ: {sex_diff:.2%}\n")
print(f"Tỷ lệ sống theo Hạng vé:\n{class_survival}\nChênh lệch Hạng 1 - Hạng 3: {class_diff:.2%}\n")

pivot = df.pivot_table(index='sex', columns='pclass', values='survived', aggfunc='mean')
print("TỶ LỆ SỐNG SÓT KẾT HỢP (GENDER x PCLASS)")
print(pivot)
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO
fare_stats = df.groupby('survived')['fare'].agg(['mean', 'median', 'std'])
print("GIÁ VÉ TRUNG BÌNH THEO TRẠNG THÁI SỐNG SÓT")
print(fare_stats)

df['fare_group'] = pd.qcut(df['fare'], q=4, labels=['Rất rẻ', 'Rẻ', 'Trung bình', 'Đắt'])
print("\nTỈ LỆ SỐNG SÓT THEO NHÓM GIÁ VÉ")
print(df.groupby('fare_group', observed=False)['survived'].mean())

corr, p_value = stats.pointbiserialr(df['survived'], df['fare'])
print(f"\nHệ số tương quan: {corr:.3f} (p-value: {p_value:.3e})")
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO
df['family_size'] = df['sibsp'] + df['parch'] + 1

def categorize_family(size):
    if size == 1:
        return 'Đi một mình (1)'
    elif 2 <= size <= 4:
        return 'Gia đình nhỏ (2-4)'
    else:
        return 'Gia đình đông (>4)'

df['family_type'] = df['family_size'].apply(categorize_family)

print("TỶ LỆ SỐNG SÓT THEO SỐ THÀNH VIÊN")
print(df.groupby('family_size')['survived'].agg(['count', 'mean']))

print("\nTỶ LỆ SỐNG SÓT THEO NHÓM GIA ĐÌNH")
print(df.groupby('family_type')['survived'].agg(['count', 'mean']))
```
